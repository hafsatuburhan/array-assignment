```python
# REAL WORLD APPLICATION OF ARRAY: Student Grade Management System
# Developed using Array Data Structure

# Array to store student grades
grades = [] # This is our array

def add_grades():
    n = int(input("How many students? "))
    for i in range(n):
        score = float(input(f"Enter score for student {i+1}: "))
        grades.append(score) # Adding to array
    print(f"\n{n} grades added to array successfully!")

def view_grades():
    print("\n--- All Student Grades ---")
    if not grades:
        print("Array is empty!")
    else:
        for i in range(len(grades)):
            print(f"Student Index {i} -> Score: {grades[i]}")

def analyze_grades():
    if not grades:
        print("No data in array!")
        return
    print("\n--- Analysis ---")
    print(f"Total students: {len(grades)}")
    print(f"Average Score: {sum(grades)/len(grades):.2f}")
    print(f"Highest Score: {max(grades)} at index {grades.index(max(grades))}")
    print(f"Lowest Score: {min(grades)} at index {grades.index(min(grades))}")

def search_grade():
    index = int(input("Enter student index to search (0 to n-1): "))
    if 0 <= index < len(grades):
        print(f"Score at index {index} is {grades[index]} -> O(1) access, this is why array is fast!")
    else:
        print("Invalid index! Array out of bounds")

# Main Menu - Run this
while True:
    print("\n==== STUDENT GRADE SYSTEM (ARRAY BASED) ====")
    print("1. Add Grades")
    print("2. View All Grades")
    print("3. Analyze (Avg, Max, Min)")
    print("4. Search by Index")
    print("5. Exit")

    choice = input("Choose: ")
    if choice == '1':
        add_grades()
    elif choice == '2':
        view_grades()
    elif choice == '3':
        analyze_grades()
    elif choice == '4':
        search_grade()
    elif choice == '5':
        break
    else:
        print("Invalid choice")
```
